# Contrasting Borehole DAS Installations at SAFOD  
## Shareable figure story: wavefield observability and monitoring sensitivity

**Purpose.** This is the lightweight, advisor-facing figure notebook for the June 2026 accelerated weight-drop experiment. It contains the current main-figure sequence, the claim supported by each figure, and the interpretation ceiling.

**Current headline.** Repeated AWD excitation produces a highly repeatable fast apparent mode on the cemented Nano fiber and a distinct slow guided mode on the Deep wireline fiber. For the selected apparent-speed observables, the Deep outbound branch reaches reliable detection at a fractional change of \(5\times10^{-3}\) (0.5%), compared with \(1\times10^{-2}\) (1%) for Nano. The Deep return branch reaches 1%, so the sensitivity advantage is **branch-specific**, not an installation-wide result.

> **Terminology:** all monitoring numbers below refer to fractional changes in the apparent along-fiber speed of the selected mode. They are not formation \(V_P\), formation \(V_S\), fault-zone stress, pore pressure, or strain.

**Question answered:** What was measured, and why should the two DAS installations not be treated as interchangeable sensors?

**Established:** The experiment contains 49 qualifying bursts and 988 individual drops. Nano is cemented behind casing over its valid interval; Deep is a wireline/hairpin installation. The source and receiver geometries, coupling, channel spacing, and usable apertures differ.

**Claim ceiling:** The installation drawing is schematic. It is not a surveyed true-depth registration or ray geometry.

## Figure 2 — Nano fast apparent mode

<img src="../figures/awd_2026/core02_nano_mode_fk_dispersion.png" width="100%">

**Question answered:** Does Nano contain a spatially ordered, directionally coherent arrival?

**Established:** A positive-slowness ridge corresponds to an apparent speed of approximately 2.95–2.99 km/s, with the clearest energy in the 30–60 Hz range.

**Claim ceiling:** This is a **coherent fast apparent mode**. The figure does not by itself establish a direct P-wave identity or a formation-\(V_P\) profile.

**Caption draft:** *Signed frequency–slowness spectrum and frequency-dependent apparent speed of the coherent Nano mode. The robust observation is a positive-slowness mode near 3 km/s with weak frequency dependence.*

## Figure 3 — Deep slow guided mode

<img src="deep_tube_record_sections.png" width="100%">

<img src="deep_sliding_fk.png" width="85%">

**Question answered:** What coherent wavefield does the Deep installation emphasize?

**Established:** A slow, coherent arrival follows both Deep legs over long along-fiber distances. Sliding frequency–slowness analysis places the primary 15–30 Hz mode near 1.4–1.56 km/s on neighboring 400 m apertures.

**Claim ceiling:** Use **slow borehole-guided mode** or **guided/tube-wave-like mode**. A physical tube-wave identity and absolute depth registration remain preliminary.

**Caption draft:** *Deep record sections and sliding signed-slowness estimates reveal a persistent slow guided mode on both legs. The repeated apparent-speed range is distinct from the fast Nano mode.*

## Figure 4 — Independent validation of the Deep mode

<img src="deep_tube_null_tests.png" width="80%">

<img src="deep_target_burst_repeatability.png" width="100%">

**Question answered:** Is the Deep trajectory a reproducible spatially ordered signal rather than a channel-order or burst-selection artifact?

**Established:** The trajectory was discovered on odd bursts and evaluated on held-out even bursts. In four leg/band tests, the unpermuted validation result exceeds all 499 channel-order permutations (\(p=0.002\)). Most target bursts recover the expected positive-slowness mode.

**Claim ceiling:** These tests establish spatial coherence and repeatability of the selected trajectory. They do not independently identify its physical phase or locate a unique generation depth.

**Caption draft:** *Held-out validation, channel-order permutation nulls, and burst-level repeatability support a real spatially ordered Deep guided-mode trajectory.*

## Figure 5 — Repeatability hierarchy

<img src="fig12_repeatability_publication.png" width="100%">

**Question answered:** Is the source–receiver system repeatable at the drop, burst, and experiment scales?

**Established:** The selected Nano waveform is substantially more coherent in the signal window than in noise. Across-burst leave-one-out stacks are highly repeatable, while individual drops show larger timing, amplitude, and waveform variability.

**Key numbers:** median signal NCC 0.889 versus noise NCC 0.288; across-burst leave-one-out median NCC 0.976; median delay \(-0.048\) ms.

**Claim ceiling:** The source is repeatable enough to recover a stable coherent moveout after stacking. It is not a perfectly repeatable individual impact source.

**Caption draft:** *Repeatability from individual drops to burst stacks. Stacking stabilizes the coherent observable, while individual-impact variability remains the principal limit on high-precision monitoring.*

## Figure 6 — Nano blind injection–recovery

<img src="nano_dvv_injection_recovery.png" width="90%">

**Question answered:** What fractional change in the selected Nano apparent-moveout observable can the present one-day design recover reliably?

**Established:** The empirical two-sided zero-injection threshold is \(3.25\times10^{-3}\) (0.325%). The smallest tested two-direction level satisfying at least 95% correct-sign detections above that threshold is \(1\times10^{-2}\) (1%).

**Claim ceiling:** This is a sensitivity floor for the selected apparent-mode estimator under this acquisition and processing design. It is not a universal DAS threshold or a formation-velocity detection limit.

**Caption draft:** *Blind injection–recovery calibration of the Nano apparent-moveout estimator. Real-data null scatter sets a 0.325% empirical threshold, and 1% is the smallest tested reliably detected change.*

## Figure 7 — Deep blind injection–recovery and installation comparison

<img src="deep_dvv_injection_recovery.png" width="100%">

| Observable | Null threshold | Reliable detection | Null scatter (MAD) |
|---|---:|---:|---:|
| Nano | \(3.25\times10^{-3}\) | \(1\times10^{-2}\) | \(1.86\times10^{-3}\) |
| **Deep outbound** | **\(1.84\times10^{-3}\)** | **\(5\times10^{-3}\)** | **\(1.20\times10^{-3}\)** |
| Deep return | \(3.05\times10^{-3}\) | \(1\times10^{-2}\) | \(1.32\times10^{-3}\) |
| Deep paired, inverse-variance | \(1.53\times10^{-3}\) | \(5\times10^{-3}\) | \(8.12\times10^{-4}\) |

**Question answered:** Does the longer Deep guided-mode trajectory improve calibrated apparent-speed sensitivity?

**Established:** The primary Deep outbound branch detects 0.5%, a factor of two below the tested Nano level. The return branch detects 1%, equal to Nano. Pairing the nearly uncorrelated legs reduces null scatter by about 1.4 times relative to outbound, but it does not cross to a lower tested detection level.

**Mechanistic result:** Deep has an 11.5-times longer regression lever arm than Nano, but its inferred per-burst timing repeatability is about 7.4 times worse. The geometric advantage therefore converts to an approximately twofold sensitivity improvement, not an 11.5-fold improvement.

**Claim ceiling:** The defensible conclusion is that the **selected Deep outbound branch** is more sensitive than the Nano apparent-moveout observable. Do not state that the Deep installation as a whole outperforms Nano.

**Caption draft:** *Blind apparent-speed injection–recovery for the Deep guided mode. The outbound branch reaches reliable detection at 0.5%, whereas the return branch reaches 1%. Combining the legs lowers the null noise floor but not the tested reliable-detection level.*

# One-slide conclusion

> **Installation controls which coherent borehole DAS mode is observable and, through the mode’s usable propagation distance and repeatability, controls monitoring sensitivity.**

- Nano emphasizes a repeatable fast apparent mode near 3 km/s.
- Deep emphasizes a distinct slow guided mode near 1.5 km/s.
- The Deep outbound trajectory has enough additional propagation leverage to improve reliable detection from 1% to 0.5%.
- Poorer burst-to-burst timing repeatability consumes most of the nominal 11.5-times lever-arm advantage.
- The result is observable- and branch-specific; it is not yet a formation-velocity or fault-zone-property measurement.

## Deliberately omitted from the main story

- **Tidal regressions and CCA:** no tidal detection is established. Tides are, at most, a detectability benchmark far below the present one-day sensitivity.
- **Raw-amplitude Nano/Deep ranking:** amplitudes are not calibrated across interrogators, installations, coupling conditions, and frequency bands.
- **Formation \(V_P\), \(V_S\), stress, pressure, permeability, or strain inference:** the calibrated quantities are apparent-mode speed changes.
- **A universal “Deep is better” statement:** only the preselected outbound branch improves on Nano under the registered decision rule.
