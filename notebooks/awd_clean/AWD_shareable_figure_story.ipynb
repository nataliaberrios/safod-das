# Contrasting Borehole DAS Installations at SAFOD
## Shareable figure story: wavefield observability and monitoring sensitivity

**Purpose.** This is the lightweight, advisor-facing figure notebook for the June 2026 accelerated weight-drop experiment. It uses ordinary Markdown image links so that GitHub renders the committed figures directly.

**Current headline.** Repeated AWD excitation produces a highly repeatable fast apparent mode on the cemented Nano fiber and a distinct slow guided mode on the Deep wireline fiber. For the selected apparent-speed observables, the Deep outbound branch reaches reliable detection at a fractional change of 0.5%, compared with 1% for Nano. The Deep return branch reaches 1%, so the sensitivity advantage is **branch-specific**, not installation-wide.

> **Terminology:** all monitoring numbers below refer to fractional changes in the apparent along-fiber speed of the selected mode. They are not formation $V_P$, formation $V_S$, fault-zone stress, pore pressure, or strain.

## Results at a glance

| Observable | Empirical null threshold | Reliable detection | Interpretation |
|---|---:|---:|---|
| Nano apparent moveout | $3.25\times10^{-3}$ (0.325%) | $1\times10^{-2}$ (1%) | Cemented-fiber fast apparent mode |
| **Deep outbound** | **$1.84\times10^{-3}$ (0.184%)** | **$5\times10^{-3}$ (0.5%)** | Primary guided-mode branch |
| Deep return | $3.05\times10^{-3}$ (0.305%) | $1\times10^{-2}$ (1%) | Independent secondary branch |
| Deep paired, inverse-variance | $1.53\times10^{-3}$ (0.153%) | $5\times10^{-3}$ (0.5%) | Lower noise floor; same tested detection level |

The outbound and return errors are nearly uncorrelated ($\rho=+0.121$). Pairing lowers null scatter, but the tested reliable-detection level remains 0.5%.

# Main figure sequence

## Figure 1 — Experiment and installation geometry

![Experiment and installation geometry](../figures/awd_2026/core01_experiment_geometry.png)

**Question answered:** What was measured, and why should the two DAS installations not be treated as interchangeable sensors?

**Established:** The experiment contains 49 qualifying bursts and 988 individual drops. Nano is cemented behind casing over its valid interval; Deep is a wireline/hairpin installation. Their coupling, channel spacing, and usable apertures differ.

**Claim ceiling:** The installation drawing is schematic, not a surveyed true-depth registration or ray geometry.

---

## Figure 2 — Nano fast apparent mode

![Nano signed frequency-slowness and dispersion](../figures/awd_2026/core02_nano_mode_fk_dispersion.png)

**Question answered:** Does Nano contain a spatially ordered, directionally coherent arrival?

**Established:** A positive-slowness ridge corresponds to an apparent speed of approximately 2.95–2.99 km/s, with the clearest energy in the 30–60 Hz range.

**Claim ceiling:** This is a **coherent fast apparent mode**. The figure does not by itself establish a direct P-wave identity or a formation-$V_P$ profile.

---

## Figure 3 — Deep slow guided mode

![Deep guided-mode record sections](deep_tube_record_sections.png)

![Deep sliding frequency-slowness result](deep_sliding_fk.png)

**Question answered:** What coherent wavefield does the Deep installation emphasize?

**Established:** A slow coherent arrival follows both Deep legs over long along-fiber distances. Sliding frequency-slowness analysis places the primary 15–30 Hz mode near 1.4–1.56 km/s on neighboring 400 m apertures.

**Claim ceiling:** Use **slow borehole-guided mode** or **guided/tube-wave-like mode**. A physical tube-wave identity and absolute depth registration remain preliminary.

---

## Figure 4 — Independent validation of the Deep mode

![Deep channel-order permutation null tests](deep_tube_null_tests.png)

![Deep target-burst repeatability](deep_target_burst_repeatability.png)

**Question answered:** Is the Deep trajectory a reproducible spatially ordered signal rather than a channel-order or burst-selection artifact?

**Established:** The trajectory was discovered on odd bursts and evaluated on held-out even bursts. In four leg/band tests, the unpermuted validation result exceeds all 499 channel-order permutations ($p=0.002$). Most target bursts recover the expected positive-slowness mode.

**Claim ceiling:** These tests establish spatial coherence and repeatability, not physical phase identity or a unique generation depth.

---

## Figure 5 — Repeatability hierarchy

![AWD repeatability hierarchy](fig12_repeatability_publication.png)

**Question answered:** Is the source-receiver system repeatable at the drop, burst, and experiment scales?

**Established:** Median signal NCC is 0.889 versus 0.288 in noise. Across-burst leave-one-out median NCC is 0.976, with median delay $-0.048$ ms. Individual drops show larger timing, amplitude, and waveform variability.

**Claim ceiling:** The source is repeatable enough to recover stable coherent moveout after stacking; it is not a perfectly repeatable individual-impact source.

---

## Figure 6 — Nano blind injection-recovery

![Nano blind injection-recovery](nano_dvv_injection_recovery.png)

**Question answered:** What fractional change in the selected Nano apparent-moveout observable can the present one-day design recover reliably?

**Established:** The empirical two-sided zero-injection threshold is $3.25\times10^{-3}$ (0.325%). The smallest tested two-direction level satisfying at least 95% correct-sign detections above that threshold is $1\times10^{-2}$ (1%).

**Claim ceiling:** This is a sensitivity floor for this selected apparent-mode estimator and experiment, not a universal DAS or formation-velocity threshold.

---

## Figure 7 — Deep blind injection-recovery and Nano comparison

![Deep blind injection-recovery](deep_dvv_injection_recovery.png)

| Observable | Null threshold | Reliable detection | Null scatter (MAD) |
|---|---:|---:|---:|
| Nano | $3.25\times10^{-3}$ | $1\times10^{-2}$ | $1.86\times10^{-3}$ |
| **Deep outbound** | **$1.84\times10^{-3}$** | **$5\times10^{-3}$** | **$1.20\times10^{-3}$** |
| Deep return | $3.05\times10^{-3}$ | $1\times10^{-2}$ | $1.32\times10^{-3}$ |
| Deep paired, inverse-variance | $1.53\times10^{-3}$ | $5\times10^{-3}$ | $8.12\times10^{-4}$ |

**Established:** The primary Deep outbound branch detects 0.5%, a factor of two below the tested Nano level. The return branch detects 1%, equal to Nano. Pairing lowers null scatter by about 1.4 times relative to outbound but does not cross to a lower tested detection level.

**Mechanistic result:** Deep has an 11.5-times longer regression lever arm than Nano, but inferred per-burst timing repeatability is about 7.4 times worse. The geometric advantage therefore converts to an approximately twofold sensitivity improvement, not an 11.5-fold improvement.

**Claim ceiling:** The **selected Deep outbound branch** is more sensitive than the Nano apparent-moveout observable. Do not state that the Deep installation as a whole outperforms Nano.

# One-slide conclusion

> **Installation controls which coherent borehole DAS mode is observable and, through the mode's usable propagation distance and repeatability, controls monitoring sensitivity.**

- Nano emphasizes a repeatable fast apparent mode near 3 km/s.
- Deep emphasizes a distinct slow guided mode near 1.5 km/s.
- The selected Deep outbound branch improves reliable detection from 1% to 0.5%.
- Poorer burst-to-burst timing repeatability consumes most of the nominal 11.5-times lever-arm advantage.
- The result is observable- and branch-specific, not yet a formation-velocity or fault-zone-property measurement.
